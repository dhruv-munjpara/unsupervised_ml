```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
```

```python
# Create dataset
data = {
    "x": [2, 3, 4, 5, 7, 9, 9, 6, 4, 3, 5, 7, 8, 6, 9, 11, 12],
    "y": [2, 9, 7, 5, 3, 1, 2, 4, 6, 8, 9, 1, 2, 3, 4, 5, 6]
}

# Create DataFrame
df = pd.DataFrame(data)
print(df)
```

```python
linked = linkage(df.values, method="ward")

# Plot dendrogram
plt.figure(figsize=(8, 5))
dendrogram(
    linked,
    labels=df.index,
    leaf_rotation=90,
    leaf_font_size=10
)

plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Euclidean Distance")
plt.grid(True)
plt.show()
```
