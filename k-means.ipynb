```python
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.cluster import KMeans

df = pd.read_csv("students_kmeans(1).csv")
df.head()

x = df[['Study_Hours_Per_Day','Exam_Score']]
print(x.head())

kmeans = KMeans(n_clusters=3,random_state=42)
kmeans.fit(x)

df['cluster'] = kmeans.predict(x)
print(df)

# literal = when we pass hardcode value 
# variable = no hardcode value , can be changed 

plt.scatter(
    df[['Study_Hours_Per_Day']],
    df['Exam_Score'],
    c=df['cluster']
)

plt.xlabel("Study Hours")
plt.ylabel('Exam_Score')
plt.title("Student_Clutering")
plt.show()
```
