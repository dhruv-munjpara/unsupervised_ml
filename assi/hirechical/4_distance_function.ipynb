# 4_distance_function

```python
import math

def distances(x1,y1,x2,y2):
    e=math.sqrt((x2-x1)**2+(y2-y1)**2)
    m=abs(x2-x1)+abs(y2-y1)
    return e,m

pairs=[((20,5),(25,8)),((15,4),(30,10)),((18,6),(28,9))]

for p1,p2 in pairs:
    e,m=distances(p1[0],p1[1],p2[0],p2[1])
    print(p1,p2)
    print("Euclidean:",round(e,2))
    print("Manhattan:",m)
```
