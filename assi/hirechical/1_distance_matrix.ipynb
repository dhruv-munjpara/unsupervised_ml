# 1_distance_matrix

```python
import pandas as pd
import numpy as np

orders={"Order":["O1","O2","O3","O4","O5","O6"],
"Delivery_Time":[20,25,15,30,18,28],
"Distance":[5,8,4,10,6,9]}

df=pd.DataFrame(orders)
X=df[["Delivery_Time","Distance"]].values

n=len(X)
matrix=np.zeros((n,n))
for i in range(n):
    for j in range(n):
        matrix[i,j]=round(np.sqrt((X[i][0]-X[j][0])**2+(X[i][1]-X[j][1])**2),2)

print(pd.DataFrame(matrix,index=df["Order"],columns=df["Order"]))
```
